## Statistiques par segment

```python
segments.groupby("Cluster")[["Age", "Total_Spent", "Nb_Achats"]].agg(["mean", "count"])
```

## Personas

Générés automatiquement à partir des statistiques par cluster ci-dessus (à affiner manuellement avec un volume de données réel).

```python
for cluster_id, group in segments.groupby("Cluster"):
    print(f"--- Segment {cluster_id} ({len(group)} client(s)) ---")
    print(f"Âge moyen : {group['Age'].mean():.0f} ans")
    print(f"Dépense moyenne : {group['Total_Spent'].mean():.0f} $")
    print(f"Fréquence d'achat moyenne : {group['Nb_Achats'].mean():.1f}")
    print(f"Catégorie préférée : {group['Categorie_Preferee'].mode().iloc[0]}")
    print()
```
