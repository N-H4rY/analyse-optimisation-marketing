# M5 — Analyse des performances des campagnes marketing

Évaluer les KPIs : impressions, clics, conversions, ROI.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
from src.data_loader import load_marketing

marketing = load_marketing()
marketing.head()
```

## Calcul des KPIs (CTR, taux de conversion, CPC, CPA)

```python
marketing["CTR"] = marketing["Clicks"] / marketing["Impressions"]
marketing["Taux_conversion"] = marketing["Conversions"] / marketing["Clicks"]
marketing["CPC"] = marketing["Budget"] / marketing["Clicks"]
marketing["CPA"] = marketing["Budget"] / marketing["Conversions"]
marketing
```

## Comparaison par canal

```python
marketing.groupby("Channel")[["CTR", "Taux_conversion", "CPC", "CPA"]].mean()
```
